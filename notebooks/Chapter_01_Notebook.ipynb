## 1.11 Functions

```python
def engagement_rate(likes, comments, views):
    """Likes plus comments, as a percentage of views."""
    return (likes + comments) / views * 100

rate = engagement_rate(200, 50, 1500)
print(round(rate, 1))
```

```python
def square_print(n):
    print(n ** 2)

def square_return(n):
    return n ** 2

result1 = square_print(4)     # shows 16, but keeps nothing
result2 = square_return(4)    # keeps 16
print(result1, result2)
```

### Section 1.11: Functions

```python
# 1
def area(width, height):
    return width * height

print(area(7, 3))

# 2
def count_word(text, word):
    return text.lower().split().count(word.lower())

print(count_word("The Dog saw the other dog", "dog"))

# 3
def percent_change(old, new):
    return (new - old) / old * 100

print(percent_change(1200, 1500))
```
